#What are variables?
Variables are symbolic names or labels that point to data objects stored in your computer's memory.

Variables allow you to store, retrieve, label, and manipulate data throughout your code without having to repeat the exact values

```python
# Storing a number, age is variable
age = 56

# Storing text (string), here user_name is variable
user_name = "Nageshwor"
```

##Uses of variables
1. Data Reusability: Store a value once and use it in multiple places without rewriting it.
2. Readability: Replace confusing, raw numbers with descriptive names (e.g., using tax_rate = 0.05 instead of just 0.05).
3. Dynamic Calculations: Hold values that change over time, like tracking a player's score or a shopping cart total.
4. Memory Management: Create a symbolic shortcut to a specific space in your computer's memory








##Rules of Python variables
1. A variable name must start with a letter or the underscore character.
2. A variable name cannot start with a number.
3. A variable name can only contain alpha-numeric characters and underscores (A-z, 0-9, and _ )
4. Variable names are case-sensitive (cast, Cast and CAST are three different variables)
5. A variable name cannot be any of the Python keywords. for (def, lamda, class, yeild)

##creating a variables inside a function,with the same name as the global variable

```python
x = "wonderful."

def myfunc():
  x = "beautiful."
  print("This place is " + x)

myfunc()

print("This place is " + x)
```

##Global keyword while creating variable

```python
def myfunc():
  global x    # If you use the global keyword, the variable belongs to the global scope

  x = "awesome"

myfunc()
print("This place is " + x)
```
