#What is Python Data Types?
Python data types are classifications that tell the Python interpreter how to store, manipulate, and operate on a specific piece of data.

##Getting the Data Type
You can get the data type of any object by using the type() function:

Class integer

```python
x = 23
print(type(x))
```

class string

```python
a = ("hello world")
print(type(a))
```

class float

```python
a = 34.45
print(type(a))
```

class list

```python
mylist = ["apple", "orange", "mango"]
print(type(mylist))
```

class set

```python
mylist = {"apple", "orange", "mango"}
print(type(mylist))
```

class tuple

```python
mylist = ("apple", "orange", "mango")
print(type(mylist))
```

class dictionary

```python
mydict = {"mango": "fruit", "cat": "animal"}
print(type(mydict))
```

class boolean

```python
a = True
print(type(a))
```
