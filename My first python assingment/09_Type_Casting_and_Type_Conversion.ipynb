#What Type casting and Type conversion?
Type conversion means changing a value from one data type to another.

For example, changing a string into an integer.

##Explicit Conversion

Explicit conversion means you manually tell Python to change the data type.

It is done by programmer.

```python
age = 20


print(age)
print(type(age))
```

```python
x = "10"

print(int(x))      # 10
print(float(x))    # 10.0
print(str(10))     # "10"
print(bool(x))     # True
```

##Implicit Conversion

Implicit conversion means Python automatically changes the data type when needed.

It is automatically done by computer.

```python
x = 10      # int
y = 2.5     # float

result = x + y

print(result)
print(type(result))
```

##Numeric → Text

Use str() to convert a number into a string.

```python
age = 20
txt= str(age)

print(txt)
print(type(txt))

```

## Text → Numeric

Use int() or float().

```python
score = "10"

print(int(score))
print(float(score))
```

##Numeric → Boolean. and boolean to numeric

Use bool().


```python
print(int(True)) #Ture converts into integer 1
print(int(False)) #False converts into integer 0
print(bool(1)) # interger converts 1 into boolean True
print(bool(0)) # interger converts 0 into boolean False
#
```

##Conversion Succeeds

A string containing a valid whole number can be converted to an integer:

```python
x = "25"
print(int(x))
```

```python
x = "10.5"
print(float(x))
```

##Conversion Produces an Error

If the string does not contain a valid number, int() cannot convert it.

```python
x = "10.5"
print(int(x))
```

```python
x = "hello"
print(int(x))
```

##How Conversion Affects Calculations and Comparisons

The data type of a value affects how Python performs calculations and comparisons. Sometimes, you need to convert a value to the correct type before using it.

```python
age = "20"  # here 20 is str, it cannot be added
print(age +5)
```

```python
age = "20"
age = int(age) # convert str to integer for calculation
print(age + 10)
```
