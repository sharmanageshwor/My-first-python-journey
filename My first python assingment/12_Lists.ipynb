#What is List in Python?
A list in Python is a collection used to store multiple values in a single variable.

Think of a list like a container or box where you can keep several items together.

Simple example

##Creating a list
We use square bracket [] to create a list.

```python
fruits = ["apple", "banana", "mango"]

print(fruits)
```

##Creating a empty list

```python
list = []
print(list)
```

##Lists Can Store Different Data Types

A list doesn't require all elements to have the same type

This list contains:

1. "Ram" → string
2. 0 → integer
3. 85.5 → float
4. True → Boolean

```python
my_list = ["apple", 20 , 85.5 , "True"]
print(my_list)
```

##List are ordered
If we add new items to a list, the new items will be placed at the end of the list.

```python
my_fruits= ["apple", "mango", "grapes"]
print(my_fruits)

mylist = ["banana", "cherry", "litchi", "papaya"]
print(mylist)
```

##List allows Duplicates


```python
nylist=["cat", "dog", "cat", "dog", "rat"]
print(nylist)
```

##List Length
To determine how many items a list has, use the len() function:



```python
nylist=["cat", "dog", "cat", "dog", "rat"]
print(len(nylist))
```

##List data types
to determine the type of list we use type() funtion:

```python
nylist=["cat", "dog", "cat", "dog", "rat"]
print(type(nylist))
```

##The list() Constructor
It is also possible to use the list() constructor when creating a new list.


##List indexing
Here python use zero based indexing which mean python start form 0 , 1 , 2 ,3 ......n.

```python
#positive indexing
game = ["football", "volleyball", "cricket", "hockey"]
print(game[0]) # 0 stands for first index and soon.
print(game[1])
print(game[2])
print(game[3])
```

##Negative indexing in python
negative index start to the end like -1, -2, -3


```python
product = ["noodles", "biscuit", "cheeseballs", "chocolate"]
print(product[-1]) # where -1 stand for first of last items
print(product[-2])
print(product[-3])
print(product[-4])
```

##Range index in list
We can specify a range of indexes by specifying where to start and where to end the range.

When specifying a range, the return value will be a new list with the specified items.



```python
students = ["raj", "mohan", "hari","mukesh", "hemant"]
print(students[2:5]) # items form index 2 to 5.
print(students[:4])  # before 4th items
print(students[2:])  # after 2 nd items.
```

## List are Changeable
we can change, add, and remove items in a list after it has been created.

##Change item value
We can change value of specific items, refer by the index of the list.


```python
marks = [20 , 40 , 45, 67, 87 ]
marks[1]= 56
print(marks)
```

##Change a range of items value

```python
marks = [20 , 40 , 45, 67, 87 ]
marks[1:3]= 56 , 99  #40 and 45 is change in 56 and 99 in output
print(marks)
```

## We can also replace less items


```python
thislist = ["apple", "banana", "cherry"]
thislist[1:3] = ["watermelon"]
print(thislist)
```

##Append item in the list
we can add an item to the end of the list, use the append() method:

```python
my_list = ["mohan", "hari", "raj" ,"saroj"]
my_list.append("hemant") # appended items always show at last of the list
print(my_list)
```

##Insert items as per index

```python
my_list = ["mohan", "hari", "raj" , "saroj"]
my_list.insert(1, "hemant")
print(my_list) # inseted item according to the index.
```

##Extend list


```python
country = ["nepal", "india", "china"]
city = ["kathmandu", "delhi", "beijing"]
country.extend(city)
print(country)
```

##Romove list

```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']
country.remove("delhi")
print(country)
```

##Remove specificed index
we use pop() fucntion to remove specified index


```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']
country.pop().  # remove last one
print(country)
```

```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']
country.pop(1)  # remove index 1 i:e india
print(country)
```

##del ()

```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']
del country[3]
print(country)
```

## clear list

```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']
country.clear()
print(country)
```

## List Slicing

Slicing allows you to get part of a list.

```python
numbers = [34 , 43, 54, 67, 78 ,12 ]
print(numbers[2:5]) # slice form middle as per index

print(numbers[:5]) # slice from the beginning

print(numbers[4:]) # slice to the end

print(numbers[:])
```

##Reverse the list

```python
numbers = [1, 2, 3, 4, 5]

print(numbers[::-1])
```

##Checking wheather an item exists


```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']

print("nepal" in country)
```

##Looping through a list


```python
country =['nepal', 'india', 'china']

for cont in country:
  print(cont)
```

##Looping through index
We can use range() and len()

```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']

for i in range(len(country)):
  print(country[i])
```

##Sorting the list
we can arrange the list in ascending order.

```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']

country.sort()
print(country)

numbers = [34 , 43, 54, 67, 78 ,12 ]
numbers.sort()
print(numbers)
```

###sorting the list decending or in reverse

```python
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']
country.sort(reverse = True)
print(country)

numbers = [34 , 43, 54, 67, 78 ,12 ]
numbers.sort(reverse = True)
print(numbers)
```

###case sensitive sort in list

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]
thislist.sort(key = str.lower)
print(thislist)
```

##Joining the lists

```python
numbers = [34 , 43, 54, 67, 78 ,12 ]
country =['nepal', 'india', 'china', 'kathmandu', 'delhi', 'beijing']

list3 = numbers + country
print(list3)
```

```python
list1 = ["a", "b" , "c"]
list2 = [1, 2, 3]

for x in list2:
  list1.append(x)

print(list1)
```
