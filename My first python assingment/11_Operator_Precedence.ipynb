#What is Operator Precedence in Python?
Operator precedence is the set of rules that tells Python which operation should be performed first when an expression contains multiple operators.

In simple words:

Operator precedence determines the order in which Python evaluates operators.

Example

```python
result = 10 + 5 * 2
print(result)
```

##Why the Order of Operations Matters

The order of operations matters because the same numbers and operators can produce different results depending on which operation is performed first.

Parentheses → Exponentiation → Multiplication/Division → Addition/Subtraction

This is similar to the mathematical rule PEMDAS/BODMAS.

##1. Parentheses have the highest priority

```python
result = (10 + 5) * 2
print(result)
```

##2. Multiplication before addition


```python
result = 10 + 2 * 3
print(result)
```

##3.Exponentiation comes before multiplication

```python
result = 2 * 3 ** 2
print(result)
```

##4.Comparison happens after arithmetic

```python
x = 10 + 5 > 12
print(x)
```

##5. Logical operators have lower precedence


```python
result = 10 > 5 and 20 > 15
print(result)
```

In one sentence:

Operator precedence is the rule Python uses to decide which operator should be evaluated first when an expression contains multiple operators.
