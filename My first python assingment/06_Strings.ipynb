#What is String in python?

A string in Python is simply a sequence of text characters (letters, numbers, or symbols) enclosed in single (') or double (") quotes.

```python
name = "Alice"
greeting = 'Hello, World!'
age_as_text = "25"
```

##Quotes inside quotes
We can use quotation marks (" or ') inside a string if they are different from the quotes used to start and end the string

```python
print("He's friend is my frind.")
print('She said, "I like Python."')
print('He is called "Micheal".')
```

##Assign string to a variable
To store a string in a variable, write the variable name, then =, then the string.


```python
name = "Nageshwor"
print(name)
```

##Multiline string
We can assign multiline string in a variable by using three quotes.

```python
a = """ Artificial Intelligence (AI) is a technology that enables computers and machines to perform tasks
 that normally require human intelligence. AI can learn from data, understand language, recognize images,
 solve problems, and make decisions. It is used in many areas such as education,
 healthcare, transportation, business, and entertainment."""
print(a)
```

##Slicing string
We can get a part of a string using slicing. Write the starting index and ending index, separated by :

```python
fame = "Must be popular"
print(fame[1:4]) # sliced for index 1 : 4
#always index starts form 0

game = "football"
print(game[2:6]) # sliced from index 2:5
```

##slice for start

```python
animal = "elephant"
print(animal[:5]) #if we dont put any number before colon, slicing begin form start
```

##slice to the end

```python
bird = "peacock"
print(bird[4:]) #if we dont put any number after colon, it slice form the last.
```

##Negative indexing

```python
animal = "elephant"
print(animal[-6:-2]) # her -6 : "e" and -2 is "n", upto "n", but no n.
```

##Modify string
There are sets of bulit in methods to modify string in python.


###upper case

```python
game = "badminton"
print(game.upper())

mycar="toyota"
print(mycar.upper())
```

###lower case

```python
animal = "THe Tiger"
print(animal.lower())
```

###Remove whitespace
to remove whitespace we use .strip() function


```python
a = "   The life is beautiful.    "
print(a.strip())#returns "The life is beautiful"
```

##Replace string

```python
a = "Your game is over"
print(a.replace("game","fame")) # here game is replace with fame
```
