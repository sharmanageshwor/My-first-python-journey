#What is String concatenation?
String concatenation means joining two or more strings together to make one string.

In Python, we use the + operator to concatenate strings.

```python
a = "I am"
b = "from Nepal."

sentence = a + b
print(sentence) # return without space
```

```python
first_name = "Nageshwor"
last_name = "Sharma"

full_name = first_name + " " + last_name # return with space

print(full_name)
```

##String format
String formatting means putting variables or values inside a string in a clean and easy way

```python
age = 25
txt = "I am Nageshwor Sharma, I am" + age # This will produce error
print(txt)
```

F-Strings
Here we have f-string which can help us to combine string and number.

```python
age =25
txt = f"I am Nageshwor Sharma. I am {age}." #This will work
print(txt)
```

##Place holder
A placeholder is a pair of curly braces {} used to show where a value should be inserted in a string.

```python
price = 25
txt = f"The price of my laptop is {price} k dollars."
print(txt)
```

##Modifiers
A modifier is used to change how a value is displayed, such as controlling decimal places or formatting numbers.

```python
price = 25
txt = f"The price of a laptop {price:.2f} dollars" # 2f means two decimals
print(txt) # if 3f means 3 decimals

txt = f"The price of a laptop {25*3}k dollars" #multiplied
print(txt)

fix = f"The price of a laptop {25+25}k dollars" # addition
print(fix)
```

##Easy way to remember:
Placeholder {} → Where to put the value

Modifier :.2f → How to display the value
