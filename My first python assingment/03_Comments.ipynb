#What is Comments in python?
A comment in Python is text within the code that the Python interpreter completely ignores during execution. It is primarily used to explain the logic, make the code more readable, or temporarily disable specific code lines while debugging.

comment begin from "#"

##Use of comments in python

1. Explain logic: Describe the why behind a complex piece of code so it is clear to read.
2. Leave reminders: Mark unfinished tasks or future updates using tags like TODO.
3. Test and debug: Turn off specific lines of code temporarily without deleting them.
4. Create documentation: Generate automatic guides and API references using specialized parsing tools

##Different Types of comment

   

1. Single-Line Comments

```python
# Calculate the area of a rectangle
length = 5
width = 3
area = length * width

print(area)
```

2. Inline Comments

```python
mycar ="porche" # porche is my favourite car
print(mycar)
```

3. Multi-Line Comments

```python
# This block of code loops
# through a list of names and
# prints a greeting for each one.
names = ["Charlie", "Jackie"]

"""
Alternatively, you can use triple quotes
to write a multi-line explanation like this
without causing any syntax errors.
"""
for name in names:
    print(f"Hello, {name}!")
```
