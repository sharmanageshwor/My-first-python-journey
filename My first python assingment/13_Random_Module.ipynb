#What is Random module?

The random module is a built-in Python module used to generate random numbers and make random choices.

You don't need to install it. You simply import it:

```python
import random
```

##random.random()
It can generate a random floating-point number between 0.0 and 1.0.

```python
import random
x = random.random()
print(x) # generate random floating point
```

##random.randrange()
It can generate a random number from a range.

```python
import random
marks=random.randrange(0,100)
print(marks)
```

##random.uniform()
It can generate a random floating-point number between two numbers.

```python
import random
x = random.uniform(1,10)
print(x)
```

##random.choice()
We can select one random item from a sequence such as a list, tuple, or string.

```python
import random

numbers = [1, 2, 3, 4, 5]

numbers = random.choice(numbers)

print(numbers)
```

choice multiple items

```python
import random
list1 = ["apple", "banana", "cherry"]
computer_choice = random.choices(list1, k=2)
print(computer_choice)
```

##random.shuffle()
we can change the original list

```python
import random

numbers = [1, 2, 3, 4, 5]

random.shuffle(numbers)

print(numbers)
```
