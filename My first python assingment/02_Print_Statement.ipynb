#Print Statement
The print() function in Python sends text or data directly to your screen or terminal window. It is a built-in function used to show results, send messages to users, or debug code during development

##Use of print statement

```python
# Printing a simple message
print("Hello, World!")

# Printing numbers and variables
age = 25
print(age)
```

```python
print("the is my world")
```

```python
print("He has", 25 , "thousand rupees in cash.")
```

##What is the purpose of displaying output?
The primary purpose of displaying output in Python is to communicate data out of the program to the user or developer. Without output capabilities, a program operates completely in the dark, computing calculations that nobody can see.



###Displaying multiple values

```python
print("hello world")
print('This is heaven') # single quotation can also be used
print("Nothing means nothing")
```

##print numbers

```python
print(6)
print(234)
print(34.45)
```

 ###Some Examples

```python
# 1. Printing multiple items with a custom separator
print("Mango", "Grapes", "Orange", sep=", ")
# Output: Mango, Grapes, Orange

# 2. Staying on the same line using the end parameter
print("Hello", end=" ")
print("Dear!")
# Output: Hello Dear!

# 3. Modern formatting using F-Strings
name = "Hari"
print(f"Welcome, {name}!")
# Output: Welcome, Hari!
```

```python
mylist = ["monkey", "donkey", "horse"]
print(mylist)   # printing variables
```
