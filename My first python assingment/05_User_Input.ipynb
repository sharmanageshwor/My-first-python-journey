#What is User input in Python?

In Python, user input is collected using the built-in input() function, which pauses program execution and waits for the user to type something into the console and press Enter.

##Why interactive programs needed?
1. Real-Time Response: They must react instantly when a user presses a key or clicks a button.
2. Event Handling: They use an event queue to process unpredictable actions from the user or operating system.
3. Continuous Running: Unlike batch jobs that run once and stop, interactive programs wait indefinitely for the next command

###for example

```python
name = input("What is your name? ")
print("Hello", name)
```

##Accepting information from the user

Python uses the input() function to accept information from the user.

```python
name = input("Enter you name: ")

print(name) # input ask user to input the name
```

##Storing user input in variables

Usually, we store user input in a variable so that we can use it later.

```python
name = input("Enter the name: ")

print("Hello", name)
print("Welcome home", name)
```

```python
city = input("Enter your city: ")

print("OH You live in", city)
```

##Converting user input into numeric types

Python provides functions such as:

int() → converts to an integer
float() → converts to a floating-point number


```python
age = int(input("Enter your age: "))

print(age + 5)
```

```python
price = float(input("Enter the price: "))

print(price)

```

##Common input-related mistakes
