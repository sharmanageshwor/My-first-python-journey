#What are Operator in Python?
Operators are special symbols or keywords used to perform operations on values and variables.

For example:

```python
a = 10
b = 5

print(a + b)
```

##Types of Operators
1. Arithmetic Operators
2. Comparison (Relational) Operators
3. Logical Operators
4. Assignment Operators
5. Bitwise Operators
6. Identity Operators
7. Membership Operators

##Arithmetic Operators

Arithmetic operators.jpg


```python
a = 18
b = 6

print(a+b) #addition
print(a-b) #subtraction
print(a*b) #Multiplication
print(a/b) #Division
print(a%b) #Modulus
print(a//b)#Quotient
print(a**b) #Exponent
```

##Equals to  and not equals to

```python
a = 6
b = 6

print(a==b) # equal to
print(a!=b) # not equal to
```

##greater than

```python
a = 8
b = 5

if a > b :
  print("a is greater than b")
```

##less than

```python
a = 5
b = 9

if a < b:
  print("a is less than b")
```

##Logical Operators
images.jpeg

```python
x = 10
print(x>5 and x < 20)
print(x<5  or x ==10)
```

```python
age = 25
has_id = True

print(age >= 18 and has_id)
```

```python
age = 25
salary = 50000

if age >= 18 and salary >= 30000:
    print("Eligible")
```

##Assignment Operators
assignment-operator-in-python.png

```python
x = 10

x += 5    # x = x + 5
x -= 2    # x = x - 2
x *= 3    # x = x * 3
x /= 2    # x = x / 2
```

```python
score = 0

score += 10
score += 20
score += 5

print(score)
```
