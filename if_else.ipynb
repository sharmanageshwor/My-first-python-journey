##Python If-else : Study Notes
###1.Making Decision with if


### An if statement tell Python to check if this condition is True or False or Just skip it.

```python
age = 17
if age >= 17:
  print("You are eligible candidate")
```

##The Fallback means

###What if the condition is False? Instead of doing nothing, we can use and else statements to catch everything else


```python
age = 17

if age>= 17:
  print("You are eligible to vote")

else:
  print("Sorry, you're too young to vote")
```

##Multlipe choices with elif
### if there is more possiblities we use elif in such condition

take a input marks from the students and classify the marks

```python
marks  = int(input("Enter the mark of student: "))

if marks >= 90:
  print("Outstanding")

elif marks >= 80:
  print("Excellent")

elif marks >= 70:
  print(" Very Good")

elif marks >= 60:
  print("Good")

elif marks >= 50:
  print("Average")

else:
  print("Not good")
```

## Combining condition(and , or , not )
we can check multiple conditions at the same time using logical operators:


```python
has_ticket = True
has_id = False

if has_ticket and has_id:
  print("welcome to the concert")
else:
  print("Access denied.")
```

##Nested if statements


```python
x = 41
if x > 10:
  print("Above ten,")
  if x > 20:
    print("and also above 20!")
  else:
    print("but not above 20.")
```

BMi calculator

```python
height = float(input("Enter the height in m: "))
weight = float(input("Enter the weight in kg: "))

bmi = weight/(height)**2

print("Your BMI is: ",bmi)

if bmi < 18.5:
  print("underweight")

elif bmi >= 18.5 and bmi < 24.9:
  print("Normal")

elif bmi >= 24.9 and bmi < 29.9:
  print("Overweight")

```

## Define leap year

```python
#leap year

year = int(input("Enter the year: "))

print(year)

if year % 4 ==0 :
  print("leap year")

elif year % 100 !=0:
  print("leap year")

elif year % 400 ==0:
  print("leap year")

else:
  print("no leap year")
```
