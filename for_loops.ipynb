##For loops
A for loops in Python is used for iterating over a sequence(such as a list, tuple, dictionary, set or string).

Use a for loop when we want to repeat a block of code a specific number of times or go through every item in a collection.



## Basic syntax


```python
for item in sequence:
  # code block to execute for each item
```

Unlike while loops (which rely on a boolean condition), a for loop automatically picks the next item from the collection until none are left.



```python
fruits = ["apple","banana","cherry"]

for fruit in fruits:
  print(fruit)
```

##Using range() for Number Sequences
When we want to repeat code a specific number of items (similar to traditional loops in other languages). we use range() function.


```python
# Print numbers form 0 to 4
for i in range(5):
  print(i)


# count by two : 2 , 4, 6, 8
for num in range(2, 10 , 2):
  print(num)
```

##Looping through other data types

```python
for letter in "Python":
  print(letter)
```

##Dictionaries(Keys and values)

```python
person = {"name": "Alex", "age": 23, "job":"Developer"}

#loop through keys
for key in person:
  print(key, person[key])

#Or use . items  () for both key and value
for keys, value in person.items():
  print(f"{key}: {value}")
```

## Loop Control Statements(break, continue, else)


Using break

```python
# using break
for number in range(1,6):
  if number ==4:
    break
  print(number)
```

##Using some handy helper functions


###enumerate(): gets both the index and the value of same time


```python
colors = ["red","green", "blue"]
for index, color in enumerate(colors):
  print(f"Index {index}:{color}")
```

##zip(): loops through multiple list simultaneously.

```python
names = ["Alex","Jack"]
scores = [85, 97]
for name, score in zip (names, scores):
  print(f"{name} scored {score}")
```

```python
## filtering and Processing data
scores =  [45, 56, 54, 23 , 98, 56]
passing_scores = []

for score in scores:
  if score >=60:
    passing_scores.append(score)

print(f"Passing scores: {passing_scores}")
```

```python
# Shopping cart total

cart_prices = [12, 99, 45, 67, 87, 45.00, 4.5]
total = 0.0

for price in cart_prices:
  total += price

print(f"Total price : Rs {total:.2f}")
```
