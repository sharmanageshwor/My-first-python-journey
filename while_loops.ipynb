## What is while loop in Python?
A while loop repeatedly executes a block of code as long as specified condition remains True.

use a while loop when we don't know in advance how many times we need to repeat an action, but we know the condition that should stop it.

##1. Basic syntax

```python
while condition:
  # code block to execute
  #increment or state change
```

##Example

```python
count = 1
while count <= 3:
  print("Count:", count)
  count+=1
```

##2. Key components


*   Condition: a boolean expression (True or False) evaluated before each iteraction.

*   Updating state: we must change the variable inside the loop so the condition eventually becomes False.


*   Infinite loops: if the condition never becomes False, the loop runs forever.






```python
x = 1
while x > 0:
  print(x) #x never change, so this runs endlessly
```

##Loop Control statements


*   break
*   continue

*   else







## Code examples
Using break:

```python
i = 1
while i <= 5:
  if i ==3:
    break # stops the loop completely when i reaches 3
  print(i)
  i += 1
```

Using continue:

```python
i = 0
while i < 5:
  i += 1
  if i == 3:
    continue # skips printing 3
  print(i)
```

Using else:

```python
count = 1
while count <= 2:
  print(count)
  count +=1

else:
  print("Loop finished successfully!")
```

Common use cases


```python
user_input = " "
while user_input.lower() != "quit":
  user_input = input("Type 'quit' to exit: ")
```

```python
age =-1

while age < 0 or age > 120:
  user_input= input("Enter your age (0 - 120): ")
  if user_input.isdigit():
    age = int(user_input)

  else:
    print("Please enter a vild number. ")

print(f"Age recorded : {age}")
```

```python
correct_pin = "4321"
attempts = 0
max_attempts = 3

while attempts < max_attempts:
  entered_pin = input("Enter 4 digit PIN: ")

  if entered_pin == correct_pin:
    print("Access granted!")
    break

  attempts += 1
  remaining = max_attempts - attempts
  print(f"Incorrect PIN. Attempts remaining: {remaining}")

else:
  print("Account locked. Too many failed attempts. ")

3


```
